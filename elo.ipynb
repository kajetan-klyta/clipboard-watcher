```python
import re
```

```python
def check_eori_format(text):
    e = re.sub(r"[\s\-\.]", "", text or "").upper()

    if re.match(r"^[A-Z]{2}[A-Z0-9]{1,15}$", e):
        return e
    else:
        return None
```

```python
result = check_eori_format('ASDFASDF')
```
